A simple rule-based chatbot built using Python,
conditional statements, and a continuous conversation loop.

## Objective

The objective of this project is to create a simple
rule-based AI chatbot that can interact with users
through predefined responses.

### Key Requirements
- Handle greetings
- Respond to basic user inputs
- Handle exit commands
- Use if-else decision-making logic
- Maintain a continuous conversation **loop**

```python
# ============================================
# Project 1: Rule-Based AI Chatbot
# DecodeLabs - Artificial Intelligence Internship
# ============================================

print("=" * 55)
print("🤖 Welcome to Robo - Rule-Based AI Chatbot!")
print("=" * 55)
print("I can chat with you using predefined rules.")
print("Type 'help' to see what I can do.")
print("Type 'bye', 'exit', or 'quit' to end the conversation.")
print("-" * 55)


# Continuous conversation loop
while True:

    # Get input from the user
    user_input = input("You: ").lower().strip()

    # Exit commands
    if user_input in ["bye", "exit", "quit"]:
        print("Bot: Goodbye! 👋 It was nice talking to you.")
        break

    # Greetings
    elif user_input in ["hi", "hello", "hey", "hii", "good morning", "good evening"]:
        print("Bot: Hello! 👋 Nice to meet you. How can I help you?")

    # Asking about the bot's name
    elif user_input in [
        "what is your name",
        "what's your name",
        "who are you"
    ]:
        print("Bot: I'm Robo 🤖, a Rule-Based AI Chatbot.")

    # Asking how the bot is doing
    elif user_input in [
        "how are you",
        "how are you?",
        "how are u"
    ]:
        print("Bot: I'm doing great! Thanks for asking. 😊")

    # Asking what the bot can do
    elif user_input in [
        "what can you do",
        "what can you do?",
        "help"
    ]:
        print("Bot: I can respond to greetings, answer basic questions,")
        print("     tell you about myself, and have a simple conversation.")

    # Asking about AI
    elif user_input in [
        "what is ai",
        "what is artificial intelligence",
        "define ai"
    ]:
        print("Bot: Artificial Intelligence (AI) is the ability of")
        print("     machines to perform tasks that normally require")
        print("     human intelligence.")

    # Asking about the project
    elif user_input in [
        "what is this project",
        "tell me about this project",
        "project"
    ]:
        print("Bot: This is a Rule-Based AI Chatbot.")
        print("     I use predefined rules and if-else logic")
        print("     to decide how to respond.")

    # Thank you responses
    elif user_input in [
        "thanks",
        "thank you",
        "thankyou"
    ]:
        print("Bot: You're welcome! 😊")

    # Positive responses
    elif user_input in [
        "good",
        "great",
        "awesome",
        "nice"
    ]:
        print("Bot: That's great to hear! 😄")

    # Negative responses
    elif user_input in [
        "bad",
        "sad",
        "not good"
    ]:
        print("Bot: I'm sorry to hear that. I hope things get better! 💙")

    # Unknown input
    else:
        print("Bot: I'm still learning! 🤔")
        print("     I don't have a rule for that yet.")
        print("     Type 'help' to see what I can understand.")

    print("-" * 55)
```
